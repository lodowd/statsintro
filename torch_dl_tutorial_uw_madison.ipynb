```python
import matplotlib.pyplot as plt

def plot_predictions(training_data=None, train_labels=None,
                     test_data=None, test_labels=None, predictions=None):
    plt.figure(figsize=[10, 7])

    # Plot the training data
    plt.scatter(training_data, train_labels, c="b", s=10, label="Training Data")

    # Plot the testing data
    plt.scatter(test_data, test_labels, c="g", s=10, label="Testing Data")

    # Plot the predictions if provided
    if predictions is not None:
        plt.scatter(test_data, predictions, c="r", s=10, label="Predictions")

    # Add legend with size specified in a dictionary
    plt.legend(prop={'size': 14})
    plt.show()
```
